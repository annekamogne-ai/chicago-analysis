This notebook uses geospatial scatter Mapbox visualizations to map crime incidents across Chicago. The goal is to identify how different crime types are distributed across the city's neighborhoods.

```python
import plotly.express as px
import pandas as pd
```

```python
crime_df=pd.read_csv('data/chicago_crime_clean.csv')
crime_df.columns.to_list()
```

```python
fig=px.scatter_mapbox(
    crime_df,
    lat='LATITUDE',
    lon='LONGITUDE',
    hover_name='PRIMARY_TYPE',
    hover_data={'COMMUNITY_AREA_NUMBER':True,'YEAR':True},
    zoom=9,
    height=600,
    width=600
)
fig.update_layout(mapbox_style='open-street-map')
fig.update_layout(title='Crime Locations in Chicago')
fig.show()
```

This scatter map shows the exact locations of crimes in Chicago. Each point represents a single crime, and hovering over a point reveals additional information such as the type of crime, the community area number, and the year it occurred. This visualization is useful for identifying specific crime incidents and understanding their geographical distribution across the city.

```python
### Heatmap (density map)
fig=px.density_map(
    crime_df,
    lat='LATITUDE',
    lon='LONGITUDE',
    z=None,
    radius=10,
    center=dict(lat=41.85,lon=-87.68),
    zoom=9,
   
    
)
fig.update_layout(title='Crime Density Heatmap in Chicago')
fig.show()
```

The density map (or heatmap) represents areas of Chicago with higher concentrations of crimes. Brighter or more intense colors indicate regions where crimes are more frequent. This type of visualization is useful for quickly identifying crime hotspots and understanding the overall spatial distribution of crime across the city.
