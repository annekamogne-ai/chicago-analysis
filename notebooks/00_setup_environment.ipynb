# 00 - Setup Environment

This notebook initializes the working environment for the *Chicago Crime Analysis* project.
The objectives are:
- Load the datasets (crime data, socio-economic indicators, schools)
- Inspect the structure and quality of the data
- Create the SQLite database for later SQL analysis


### Import Libraries

```python
import pandas as pd
import sqlite3
%load_ext sql
```

### Load datasets

##### Loading the datasets

We load the three main datasets:
- Crime dataset
- Socio-economic indicators
- Schools dataset


```python
crimes=pd.read_csv('data/chicago_crime.csv')
socio=pd.read_csv('data/chicago_census.csv')
schools=pd.read_csv('data/chicago_school.csv')
crimes.head()
#socio.head()
#crimes.head()
```

```python
# Let's take a first idea of the size of the load dataset
print(f"Crimes dataset shape:{crimes.shape}\nSocio dataset shape:{socio.shape}\nSchools dataset shape:{schools.shape}")
```

### Creating the SQLite database

We create the `chicago.db` database inside the `database/` folder.  
All datasets will be stored as SQL tables.


```python
# Connect to SQLite
con=sqlite3.connect('database/chicago.db')
# Let's first connect to our database
%sql sqlite:///database/chicago.db

# Save dataframes as SQL tables
crimes.to_sql('crimes_data',con,if_exists='replace')
socio.to_sql('socio_data',con,if_exists='replace')
schools.to_sql('schools_data',con,if_exists='replace')
```

```python
# Let's check our tables names
%sql SELECT name FROM sqlite_master WHERE type='table'
```
