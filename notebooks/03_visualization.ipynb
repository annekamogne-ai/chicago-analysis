In the following sections, I will explore the socio-economic landscape of Chicago through a series of visualizations. Using bar charts, heatmaps, and scatter plots, I aim to highlight key patterns and relationships across different community areas. These visual tools will help provide an initial understanding of how income levels, hardship index, population characteristics, and other socio-economic indicators vary throughout the city.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
socio_df=pd.read_csv('data/chicago_census_clean.csv')
socio_df.columns.to_list()
```

### Top Communities: Highest Poverty Rate

```python
# Top 10 poverty
top_poverty=socio_df.nlargest(10,'pct_below_poverty')
plt.figure(figsize=(20,14))
sns.barplot(
    data=top_poverty,
    x='pct_below_poverty',
    y='community_name'
)
plt.title('Top communities with Highest Poverty Rate',fontsize=20)
plt.xlabel('Poverty Rate (%)',fontsize=15)
plt.ylabel('communities',fontsize=15)
plt.show()
```

### Scatter plot: Income vs Hardship Index

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=socio_df,
    x='income_per_capita',
    y='hardship_index'
)
plt.title('Relationship between Per  Capita Income and Hardship Index')
plt.xlabel('Per Capita Income')
plt.ylabel('Hardship Index')
plt.show()
```

```python
socio_df[['income_per_capita','hardship_index']].corr()
```

The scatter plot shows a clear negative relationship between per capita income and the hardship index.
This means that
Communities with higher per capita income tend to have a lower hardship index, indicating better living conditions.
Conversely, communities with lower per capita income tend to experience higher hardship, reflecting greater socio-economic difficulties.
This negative correlation confirms that income is an important factor influencing overall hardship in Chicago’s communities.

### Histogram: Per Capita Distribution

```python
plt.figure(figsize=(10,6))
sns.histplot(
    data=socio_df,
    x='income_per_capita',
    bins=20,
    kde=True
)
plt.title('Distribution of Per Capita')
plt.xlabel('Per Capita Income (USD)')
plt.ylabel('Frequency')
plt.show()
```

The histogram shows the distribution of per capita income across Chicago’s communities.
Most communities have per capita incomes between roughly $10,000  and  $30,000, indicating that the majority of the population lives in lower- to middle-income areas.

There are a few communities with much higher incomes, creating a right-skewed distribution.
This skew indicates income inequality, with a small number of wealthier communities standing out compared to the majority.

Overall, the histogram gives a clear picture of the economic disparities across Chicago’s neighborhoods.

### Vulnerable Population Top 10 Communities

```python
# Top 10 vulnerable (young + old)
top_vulnerable=socio_df.nlargest(10,'pct_young_old')
plt.figure(figsize=(10,6))
sns.barplot(
    data=top_vulnerable,
    x='pct_young_old',
    y='community_name'
)
plt.title('Top 10 Communities with Highest Percentage of Vulnerable Populations')
plt.xlabel('Percentage (%)')
plt.ylabel('Community')
plt.show()
```

### Correlation Heatmap

```python
plt.figure(figsize=(10,8))
sns.heatmap(
    socio_df.corr(numeric_only=True),
    cmap='coolwarm',
    annot=True
)
plt.title('Correlation Heatmap of Socio-Economic Indicators')
plt.show()
```

In this notebook, we explored the socio-economic conditions of Chicago’s communities through several key visualizations.
The bar plots highlighted the communities most affected by poverty, and vulnerable populations, while the scatter plot and histogram revealed clear patterns in income distribution and its relationship with the hardship index.
Finally, the correlation heatmap provided a comprehensive overview of the relationships among the socio-economic indicators.

Overall, these visualizations give valuable insights into economic disparities across Chicago, setting a strong foundation for further analysis and integration with crime data in subsequent notebooks and the eventual interactive dashboard
