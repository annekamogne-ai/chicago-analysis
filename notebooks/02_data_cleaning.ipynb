#### In this notebook, I perform an initial cleaning of both thr socio-economic dataset and the crime dataset from the city of Chicago.
The goal of this step is to ensure consistent formatting, handle missing values, and prepare the data for further analysis.
This preprocessinh will enable two subsequent notebooks, one dedicatde to exploring the socio-economic indicators through statistical visualizations, and another focused on geospatial analysis of crime patterns across Chicago.These exploring analysis will ultimately support the development if an iteractive dashboard that highlights key insights about the city.

```python
import pandas as pd
```

### Socio-economic indicator dataset

```python
df=pd.read_csv('data/chicago_census.csv')
df.head()
```

```python
df.info()
```

```python
# We notice only two missing values
df.dropna(inplace=True)
```

```python
df.info()
```

```python
df.rename(columns={
    'PERCENT_OF_HOUSING_CROWDED':'pct_crowed_housing',
    'PERCENT_HOUSEHOLDS_BELOW_POVERTY':'pct_below_poverty',
    'PERCENT_AGED_16__UNEMPLOYED':'pct_unemployed',
    'PERCENT_AGED_25__WITHOUT_HIGH_SCHOOL_DIPLOMA':'pct_without_hs',
    'PERCENT_AGED_UNDER_18_OR_OVER_64':'pct_young_old',
    'PER_CAPITA_INCOME':'income_per_capita',
    'HARDSHIP_INDEX':'hardship_index',
    'COMMUNITY_AREA_NUMBER':'community_number',
    'COMMUNITY_AREA_NAME':'community_name'
},inplace=True)
df.columns.to_list()
```

```python
df.to_csv('data/chicago_census_clean.csv',index=False)
```

### Crime dataset

```python
df1=pd.read_csv('data/chicago_crime.csv')
df1.head()
```

```python
df1.info()
```

```python
# We notice many missing values
df1.dropna(inplace=True)
df1.info()
```

```python
# Convert DATE to datetime
df1['DATE']=pd.to_datetime(df1['DATE'])
df1['DATE'].dtype
```

```python
df1.to_csv('data/chicago_crime_clean.csv',index=False)
```
